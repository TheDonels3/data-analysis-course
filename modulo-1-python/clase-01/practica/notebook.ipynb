## Ejercicio 3 — Casting para el análisis

Muchos sistemas de scraping o formularios web devuelven todos los datos como strings, incluso los números. Antes de poder operar con esos datos, tiens que convertirlos al tipo correcto.

**Consigna:** El sistema te entregó estos datos de un producto como strings. Conviertelos al tipo adecuado para poder calcular el precio con descuento y el IVA.

```python
# Datos del sistema externo
precio_str = "15990"
descuento_str = "0.15"
cantidad_str = "4"
tiene_iva_str = "True"

# 1. Convertir los datos
precio = int(precio_str)
descuento = float(descuento_str)
cantidad = int(cantidad_str)
tiene_iva = tiene_iva_str == "True"

# 2. Aplicar el descuento
precio_final = precio * (1 - descuento)

# 3. Sumar el IVA si corresponde
if tiene_iva:
    precio_final = precio_final * 1.21

# 4. Calcular el total
total = precio_final * cantidad

print(f"Total: ${total:.2f}")
```

## Ejercicio 4 — Reporte de ventas básico

Tu jefa te pide que armes un mini-reporte de ventas del día para enviarlo por mensaje. Tiene que ser legible y estar en una sola variable.

**Consigna:** Usa las variables dadas para construir un reporte en varias líneas usando \n dentro de un string, o un f-string multilínea con """. El reporte debe incluir todos los campos y calcular el promedio de ventas.

```python
fecha = "2024-11-14"
vendedor = "Carlos Méndez"
ventas = [45000, 32000, 67500, 28000, 91000]
producto_estrella = "Auriculares Bluetooth"

# Calcular el total y el promedio
total_ventas = sum(ventas)
promedio_ventas = total_ventas / len(ventas)

# Construir el reporte
reporte = f"""Reporte de ventas — {fecha}
Vendedor: {vendedor}
Total ventas: ${total_ventas}
Promedio por venta: ${promedio_ventas}
Producto estrella: {producto_estrella}"""

print(reporte)
```

## Ejercicio 7 — Detective de tipos

En proyectos reales de data analisis, los datasets suelen tener columnas con tipos mezclados o inesperados. Saber detectar y manejar esto a mano, te da una comprensión mucho más sólida de lo que las herramientas hacen por ti.

**Consigna:** Tienes estos valores tal como podrían llegar de un CSV real. Para cada uno:


- Determina qué tipo tiene actualmente con **type()**
- Determina qué tipo debería tener para análisis
- Conviertelo (si es posible) y maneja los casos problemáticos.

```python
from datetime import datetime

# Valores originales
valor_a = "42"
valor_b = "3.14"
valor_c = "True"
valor_d = ""
valor_e = "N/A"
valor_f = "  150  "
valor_g = "$1,250.50"
valor_h = "01/03/2024"

# 1. Mostrar los tipos originales
print("=== TIPOS ORIGINALES ===")
print(type(valor_a))
print(type(valor_b))
print(type(valor_c))
print(type(valor_d))
print(type(valor_e))
print(type(valor_f))
print(type(valor_g))
print(type(valor_h))

# 2. Convertir los valores al tipo adecuado
a = int(valor_a)
b = float(valor_b)
c = valor_c == "True"
d = None if valor_d.strip() == "" else valor_d
e = None if valor_e.strip() == "N/A" else valor_e
f = int(valor_f.strip())
g = float(valor_g.replace("$", "").replace(",", ""))
h = datetime.strptime(valor_h, "%d/%m/%Y")

# 3. Mostrar los valores convertidos y sus tipos
print("\n=== VALORES CONVERTIDOS ===")
print(f"Valor A: {a} | Tipo: {type(a)}")
print(f"Valor B: {b} | Tipo: {type(b)}")
print(f"Valor C: {c} | Tipo: {type(c)}")
print(f"Valor D: {d} | Tipo: {type(d)}")
print(f"Valor E: {e} | Tipo: {type(e)}")
print(f"Valor F: {f} | Tipo: {type(f)}")
print(f"Valor G: {g} | Tipo: {type(g)}")
print(f"Valor H: {h} | Tipo: {type(h)}")
```

Para **valor_g** y **valor_f**, vas a necesitar métodos de strings como **.strip()**, **.replace()**. Investigalos.
